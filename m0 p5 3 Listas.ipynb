## Listas
[al indice](#Contenidos)  
Se trata de otro de los tipos de datos de Python más usados. Dentro de las colecciones, que veremos más adelante, la lista es la colección que normalmente se le da más uso. **Nos permiten almacenar conjuntos de variables u objetos**, y son elementos de lo más versátiles puesto que podemos almacenar objetos de distintos tipos, modificarlos, eliminarlos, meter listas dentro de listas... Sus dos caractrísticas principales son:
* **Mutables**: una vez se ha creado la lista, se puede modificar
* **Ordenada**: Los elementos tienen un cierto orden, lo que nos permite acceder al elemento que queramos teniendo en cuenta tal orden

En cuanto a su sintaxis, cuando declaremos la lista simplemente hay que separar cada elemento con comas, y rodearlo todo con corchetes.

```python
#Lista de números
numeros=[1,3,7,9]
print(numeros)
print(type(numeros))
```

```python
#Lista de strings
lista_strings = ["Esto", "es", "una", "lista"]
print(lista_strings)
print(type(lista_strings))
```

```python
#Lista de booleanos
lista_booleanos =[True,False,not False, True or False, False and False]
print(lista_booleanos)
print (type(lista_booleanos))
```

```python
#Lista de tipos mezclados
lista_mix = ["Esto es", 13,-7,True,False,"lista"]
print(lista_mix)
print (type(lista_mix))

#lista dentro de listas
listas = ["Kaixo", "7", 7, 21, lista_mix]
print(listas)
print(type(listas))
```

```python
len(listas) #longitud, number of items in a container
```

```python
len(lista_mix)
```

```python
concatenación = lista_mix + listas #colecciones ordenadas
print(concatenación)
print(type(concatenación))
```

```python
concatenación2 = listas + lista_mix 
print(concatenación2)
print(type(concatenación2))
```

```python
type(lista_mix)
```

Calcular la longitud de la misma mediante el método *built-in* ya visto: `len()`

Accedemos a los elemenos de la lista mediante corchetes `[]`

**Importante**. El primer elemento es el 0

```python
lista_ejemplo = ["posición cero", "posicion 1", 2, 3, "4"]
print(lista_ejemplo[0]) #el primer elemento
print(lista_ejemplo[-1]) #el último elemento
print(lista_ejemplo[-2]) #el penúltimo elemento (recorriendo la lista en sentido inverso)
print(lista_ejemplo[21]) #error fuera de rango
```

### Metodos en Listas
Para el tipo de objeto lista, también hay una serie de métodos catacterísticos que nos permiten operar con ellas: añadir valores, quitarlos, indexado, filtrado, etc... En [este enlace](https://www.w3schools.com/python/python_ref_list.asp) puedes encontrar todos los métodos que podrás usar con listas.

```python
dias = ["Lunes", "Martes", "miércoles"]
#añadir elementos a la lista
dias.append("juernes")
print(dias)
```

```python
#añadir elementos en una posición determinada
dias.insert(3,"jueves")
print (dias)
```

```python
#quitar elementos
#dias.clear() #vacía la lista
#print (dias)
```

```python
#quitar un elemento a partir de su índice o posición
dias.pop(4)
print (dias)
```

```python
#quitar un elemento por su valor
dias.remove("jueves")
print (dias)
```

```python
## generando errores por valores que no existen 
dias.remove("sábado")
print (dias)
```

```python
## generando errores por índice o posición que no existen 
dias.pop(8)
print (dias)
```

```python
#ordenar la lista
dias = ["lunes", "martes", "miércoles", "jueves", "viernes", "sábado", "domingo"]
dias.sort()
print (dias)
```
