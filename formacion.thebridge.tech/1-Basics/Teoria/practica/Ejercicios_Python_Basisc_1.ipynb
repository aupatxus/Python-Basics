## Ejercicio 1

1. Crea dos variables numericas: un `int` y un `float`
2. Comprueba sus tipos
3. Sumalas en otra nueva
4. ¿De qué tipo es la nueva variable?
5. Elimina las dos primeras variables creadas

```python
print("*"*32)
var1 = 7 #int
var2 = 3.1416 #float
print("\nLa varible var1 es del tipo",type(var1), "Y su valor es", var1)
print("\nLa varible var2 es del tipo",type(var2), "Y su valor es", var2)
var3 = var1 + var2
print("\nLa varible var3 es del tipo",type(var3), "Y su valor es", var3, "que se corresponde con la suma de las anteriores.")

del var1,var2
print("\nEliminadas con 'DEL' las dos primeras variables creadas.\n")
print("*"*32)
```

Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

radianes = grados*(pi/180)

```python
print("*"*32)

print("\nPrograma para pasar de Grados a Radianes.")
grados = float(input("\nIntroduce los Grados:").replace(",","."))
pi = 3.1416
radianes = grados*(pi/180)
print("\nRadianes = Grados * (Pi / 180)")
print(grados,"Grados corresponden a", radianes, "Radianes\n")

print("*"*32)
```

Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`

```python
print("*"*32)

print("\nPrograma para calcular el area de un paralelogramo (base x altura).")
base = float(input("\n Introduce la base:").replace(",","."))
altura = float(input("\n Introduce la altura:").replace(",","."))
print("\nEl área de un paralelogramo con base", base,"y altura",altura,"es de", base*altura, "unidades cuadradas.\n" )

print("*"*32)
```
