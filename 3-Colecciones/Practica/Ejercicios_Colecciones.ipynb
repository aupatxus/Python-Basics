## Ejercicio 3
Crea una lista nueva con todas las combinaciones de las siguientes dos listas:
> ```ejer_3_1 = ["Hola", "amigo"]```
>
> ```ejer_3_2 = ["Que", "tal"]```

Obten el siguiente output:

```['Hola Que', 'Hola tal', 'amigo Que', 'amigo tal']```

````python
ejer_3_1 = ["Hola", "amigo"]
ejer_3_2 = ["Que", "tal"]

resultado = []

#recorriendo las listas empleando índices
for indice1 in range(len(ejer_3_1)):
        for indice2 in range(len(ejer_3_2)):
            valores_combinados = ejer_3_1[indice1] + " " + ejer_3_2[indice2]
            resultado.append(valores_combinados)

print(resultado)
````

````python
ejer_3_1 = ["Hola", "amigo"]
ejer_3_2 = ["Que", "tal"]


resultado = []

for palabra1 in ejer_3_1:
    for palabra2 in ejer_3_2:
        valores_combinados = palabra1 + " " + palabra2
        #print(valores_combinados)
        resultado.append(valores_combinados)

print(resultado)
````

## Ejercicio 5
Dada la siguiente lista, elimina todos los valores iguales a 3
> ```ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]```

TIP: No intentes eliminar elementos sobre la lista que estás iterando. Haz una copia con `ejer_5.copy()`.

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]
print(ejer_5)

copia_ejer_5 = ejer_5.copy()

for elemento in copia_ejer_5:
    if elemento == 3:
        ejer_5.remove(3)

print(ejer_5)
````

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]
print(ejer_5)

indice = 0

while indice < len(ejer_5):
    if ejer_5[indice] == 3: 
        ejer_5.pop(indice) # borramos el contenido según su posición
    else:
        indice = indice + 1

print(ejer_5)
````

## Ejercicio 11
Convierte la siguiente lista en un diccionario
> ```ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]```

TIP: Puedes usar dos variables de control en un bucle `for`. Una por cada valor de las tuplas.

````python
ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]

resultado = {}

for letra, numero in ejer_11:
    # print ("Letra", letra, "Número", numero)
      
    if letra in resultado: # Si la letra ya estaba en el diccionario, añadimos el número en su lista
        resultado[letra].append(numero)
    else: # la letra aparece por primera vez, por lo que creamos una nueva lista con el número 
        resultado[letra] = [numero]

print(resultado)
````

## Ejercicio 15
Multiplica todos los valores del diccionario
