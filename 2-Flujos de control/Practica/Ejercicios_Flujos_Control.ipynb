## Ejercicio 1
Declara una variable numérica que será una hora cualquiera del día. Implementa la siguiente casuística:

* Si es entre las 0 y las 8, print "Durmiendo"
* Si es entre las 9 y las 18, print "Trabajando"
* Si es entre las 19 y las 21, print "Clase"
* Si es entre las 22 y las 24, print "Descanso"
* En cualquier otro caso, print "Transporte o error"

```python
hora = 99 # Declaración de una variable numérica que será una hora cualquiera del día

if hora >= 0 and hora <= 8: # Si es entre las 0 y las 8, print "Durmiendo"
    print("Durmiendo")
else:
    if hora >= 9 and hora <= 18:
        print("Trabajando") # Si es entre las 9 y las 18, print "Trabajando"
    else:
        if hora >= 19 and hora <= 21: # Si es entre las 19 y las 21, print "Clase"
            print("Clase")
        else:
            if hora >= 22 and hora <= 24: # Si es entre las 22 y las 24, print "Descanso"
                print("Descanso")
            else:
                print("Transporte o error") # En cualquier otro caso, print "Transporte o error"

```

```python
hora = 99 # Declaración de una variable numérica que será una hora cualquiera del día

match hora:
    case _ if hora >= 0 and hora <= 8:
        print("Durmiendo")
    case _ if hora >= 9 and hora <= 18:
        print("Trabajando")
    case _ if hora >= 19 and hora <= 21:
        print("Clase")
    case _ if hora >= 22 and hora <= 24:
        print("Descanso")
    case _:
        print("Transporte o error")
```

## Ejercicio 9
En este ejercicio vamos a crear un pequeño juego. Se trata de intentar adivinar un numero del 1 al 5. Tenemos dos intentos para acertar. Pasos a seguir:

1. Ya viene implementado con la librería `random` cómo obtener un número aleatorio del 1 al 5
2. Tendrás que declarar en una variable el numero de vidas, y mediante un bucle while, comprobar que todavia quedan vidas.
3. Dentro del bucle, obtener el valor del usuario y comprobar si es ese el numero a adivinar. Si no, actualizar las vidas.
4. Si acertamos, salimos del bucle e imprimimos por pantalla "You win". Y si perdemos también salimos del bucle, pero en este caso imprimimos por pantalla "You lose".

TIP: te puede resultar útil usar la sentencia `else` cuando acabe el bucle `while`. Lo que haya dentro de ese `else` se ejecutará una vez acabe la ejecución del `while`. Lo podrás usar para cuando pierdas.

```python
import random

numero_aleatorio = random.randint(1, 5) # número aleatorio del 1 al 5
vidas = 2

print("Adivina un número elegido al azar")
print("*"*33)

while vidas > 0: # bucle while para comprobar que todavía quedan vidas
    print("Dispones de", vidas, "vidas para adivinar un número al azar del 1 al 5: ...")
    propuesta_num = int(input("¿Adivina cuál es el número que estoy pensando?"))
    
    if propuesta_num == numero_aleatorio:
        print("You win")
        break
    else:
        vidas = vidas - 1
        if vidas > 0:
            print(propuesta_num, "no es el número en que estoy pensando. Inténtalo otra vez, te quedan", vidas, "vidas.")
        else: # no quedan vidas para más intentos
            print("You lose")


print("El número que había elegido al azar era el", numero_aleatorio)
```
