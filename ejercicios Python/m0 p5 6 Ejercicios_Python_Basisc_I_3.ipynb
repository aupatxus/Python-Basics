## Ejercicio 1
Tenemos las siguientes variables:

```Python
A = 4
B = "Text"
C = 4.1
```

Comprueba:
1. Si A y B son equivalentes
2. Si A y C NO son equivalentes
3. Si A es mayor que C
4. Si C es menor o igual que A
5. Si B NO es equivalente a C

````python
A = 4
B = "Text"
C = 4.1

if A==B:
    print("A y B son equivalentes")
else:
    print("A y B NO son equivalentes")

if A!=C:
    print("A y C NO son equivalentes")
else:
    print("A y C son equivalentes")

if A>C:
    print("A es mayor que C")
else:
    print("A NO es mayor que C")

if C<=A:
    print("C es menor o igual que A")
else:
    print("C NO es menor o igual que A")

if B!=C:
    print("B NO es equivalente a C")
else:
    print("B es equivalente a C")




````

## Ejercicio 3
Mismo programa que en 2, pero en esta ocasión tienen que ser tres inputs y dos salidas. Una de las salidas que nos indique si todos son iguales, y la otra si al menos dos inputs sí que lo son

````python
entrada1 = input("Introduce el primer input")
entrada2 = input("Introduce el segundo input")
entrada3 = input("Introduce el tercer input")

if entrada1 == entrada2 and entrada2 == entrada3:
    print("Esos 3 inputs son todos iguales")
else:
    print("Esos 3 inputs NO son todos iguales")

if entrada1 == entrada2 or entrada2 == entrada3 or entrada1 == entrada3:
    print("2 de los 3 inputs son iguales")
else:
    print("NO hay 2 de los 3 inputs iguales")
````

## Ejercicio 4
Razona sin ejecutar código el output que obtendremos de las siguientes sentencias
1. True and True and False
2. not ((True or False) and (True or False))
3. (False or False or False or True) and False
4. not False and True and not True

````python
# True and True and False = FALSO
# 1 and 1 and 0 = 0
var1 = (True and True and False)
print(var1)

# not ((True or False) and (True or False)) = FALSO
# not ((1 or 0) and (1 or 0))
# not ((1) and (1))
# not (1) = 0
var2 = (not ((True or False) and (True or False)))
print(var2)

# (False or False or False or True) and False = FALSO
# (0 or 0 or 0 or 1) and 0
# (0 or 0 or 0 or 1) and 0
# (1) and 0 = 0
var3= (False or False or False or True) and False
print(var3)

# not False and True and not True = FALSO
# not 0 and 1 and not 1
# 1 and 1 and 0 = 0
var4=not False and True and not True
print(var4)
````
