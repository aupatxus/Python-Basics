## Ejercicio 1
Eleva todos los elementos de la lista al cuadrado
> ```ejer_2 = [1,2,3,4,5]```

`````python
ejer_2 = [1,2,3,4,5]
print(ejer_2)

for i in ejer_2:
    print("\nPosición:",i-1,"; Su contenido antes de elevar al cuadrado:", ejer_2[i-1]) #desde el cero, que es la primera posición
    # Un número al cuadrado es el resultado de multiplicar ese número por sí mismo.
    # Elevar al cuadrado es una operación de potencia donde el exponente es siempre 2
    
    # elev_cuadrado = ejer_2[i-1] * ejer_2[i-1]
    elev_cuadrado = ejer_2[i-1] ** 2
    print(ejer_2[i-1], "elevado al cuadrado es:", elev_cuadrado)
    ejer_2[i-1] = elev_cuadrado #guardo el nuevo resultado en la lista

print("\nLa lista ha actualizado sus elementos elevados al cuadrado")
print(ejer_2)
`````

## Ejercicio 13
Escribe una función que compruebe si un DNI es válido, devolviendo `True` en caso afirmativo.  
En primer lugar, el DNI debe tener entre 7 y 8 dígitos numéricos.  
Para saber si la letra del DNI es correcta puedes utilizar el siguiente fragmento de código:
>```Python
>DNI= 55555555
>palabra='TRWAGMYFPDXBNJZSQVHLCKE'
>print('La letra del DNI es: ', palabra[DNI%23])
>````


`````python
DNI= 55555555
palabra='TRWAGMYFPDXBNJZSQVHLCKE'
print('La letra del DNI es: ', palabra[DNI%23])
`````

`````python
def validar_dni():
    dni_completo = input("Introduce tu DNI (números y letra, sin espacios ni guiones): ").upper()
    
    # La letra es el último carácter
    letra_ingresada = dni_completo[-1]
    
    if letra_ingresada.isdigit():
        print("\nERROR: El DNI introducido no contiene letra.")
        return False
        
    digitos_texto = dni_completo[:-1] # para extraer todo lo anterior (los números) y descartar la letra.

    
    # Validación de errores en la parte numérica, que contenga solo dígitos y presente entre 7 y 8 caracteres
    # if 7 <= len(digitos_texto) <= 8 and digitos_texto.isdigit():
    # if len(digitos_texto) < 7 or len(digitos_texto) > 8:
    # if not digitos_texto.isdigit() or not (7 <= len(digitos_texto) <= 8):
    # Comprobamos que el tamaño sea correcto Y que sean solo números
    
    if len(digitos_texto) < 7 or len(digitos_texto) > 8:
        print("ERROR: El DNI debe tener entre 7 y 8 números seguidos de una letra.")
        return False

    if not digitos_texto.isdigit():
        print("ERROR: El DNI debe tener entre 7 y 8 números y no letras.")
        return False
        
        
    # Convertir la parte numérica a entero para la operación matemática
    dni_numerico = int(digitos_texto)
    
    # Calcular la letra correcta usando el fragmento de código proporcionado
    palabra = 'TRWAGMYFPDXBNJZSQVHLCKE'
    letra_correcta = palabra[dni_numerico % 23]
    
    # Comprobación de si la letra introducida coincide con la calculada
    if letra_ingresada == letra_correcta:
        print("\nEl DNI es válido. La letra",letra_ingresada,"es correcta.")
        return True
    else:
        print("\nEl DNI NO es válido. La letra ingresada fue", letra_ingresada,", pero debería ser",letra_correcta,".")
        return False

# Llamada a la función
validar_dni()
`````
