## PARTE I

En este ejercicio debes simular un chatbot al que le haremos pedidos de pizzas.

1. El chatbot tiene que saludar con un: "Buenas tardes, bienvenido al servicio de pedido online, el precio de nuestra pizzas es de 8,95 para la familiar o de 9,90 para la mediana ¿Cuántas pizzas  familiares desea?"
2. El ususario tiene que introducir un número de pizzas en una variable llamada 'pizz_familiar'
3. El chatbot debe responder: "Estupendo, ¿cuántas pizzas medianas desea?" y guardar el número de pizzas medianas en otra variable 'pizz_mediana'
4. Además debe responder del chatbot: "Estupendo, se están preparando 'pizz' pizzas familiares y 'pizz_mediana' pizzas medianas. Digame su dirección"
5. El ususario tiene que introducir una direccion en formato String en otra variable llamada 'direcc'
6. El programa debe calcular el monto total
7. Respuesta final del chatbot: "Le mandaremos las '(número total de pizzas)' pizzas a la dirección 'direcc'. Serán <precio total a pagar en euros>. Muchas gracias por su pedido."
         


```python
print("-" * 30)
#El chatbot tiene que saludar con un: "Buenas tardes, bienvenido al servicio de pedido online, el precio de nuestra pizzas es de 8,95 para la familiar o de 9,90 para la mediana ¿Cuántas pizzas familiares desea?"
print("\nBuenas tardes, bienvenido al servicio de pedido online. El precio de nuestra pizzas es de 8,95 para la familiar o de 9,90 para la mediana. ")
#El ususario tiene que introducir un número de pizzas en una variable llamada 'pizz_familiar'
pizz_familiar = int(input("\n¿Cuántas pizzas familiares desea?"))
#El chatbot debe responder: "Estupendo, ¿cuántas pizzas medianas desea?" y guardar el número de pizzas medianas en otra variable 'pizz_mediana'
pizz_mediana = int(input("\nEstupendo, ¿cuántas pizzas medianas desea?"))
#Además debe responder del chatbot: "Estupendo, se están preparando 'pizz' pizzas familiares y 'pizz_mediana' pizzas medianas. Digame su dirección"
print("\nEstupendo, se están preparando", pizz_familiar,"pizzas familiares y", pizz_mediana, "pizzas medianas.")
#El ususario tiene que introducir una direccion en formato String en otra variable llamada 'direcc'
direcc = input("\nDígame su dirección")
#El programa debe calcular el monto total
totalpedido = 8.95*pizz_familiar + 9.90*pizz_mediana
totalpizzas = pizz_mediana + pizz_familiar
#Respuesta final del chatbot: "Le mandaremos las '(número total de pizzas)' pizzas a la dirección 'direcc'. Serán . Muchas gracias por su pedido."
print("\nLe mandaremos las", totalpizzas, "pizzas a la dirección",direcc,".")
print("\nSerán",totalpedido,"€. Muchas gracias por su pedido.")
print("-" * 30)
```

## Parte II

Crea un programa que realice las siguientes acciones:

1. Solicite al usuario que introduzca 3 números enteros separados por comas (por ejemplo: 1,2,3).
2. Convierta esa entrada en una lista de números enteros.
3. Utilizando funciones built-in, determine el número máximo, el número mínimo y la suma total de los números introducidos.
4. Determine la verdad o falsedad de la siguiente afirmación: "El número máximo menos el número mínimo es igual a la suma total dividida por 5".
5. Muestre los resultados utilizando print.

```python
print("-" * 32)
#Solicite al usuario que introduzca 3 números enteros separados por comas (por ejemplo: 1,2,3).
entrada=input("\nIntroduce 3 números enteros separados por comas (por ejemplo: 1,2,3)")

#Convierta esa entrada en una lista de números enteros.

lista_enteros = list(map(int, entrada.split(",")))
#entrada.split(","): Corta el texto cada vez que encuentra una coma y lo convierte en una lista de strings: ['1', '2', '3']
#.map(int, ...): Transforma cada uno de los textos individuales ('1', '2', '3') en un número entero (1, 2, 3)
#list(...): Lo convierte para guardarlo en formato de lista [1, 2, 3]

#lista_enteros2 = [int(numero) for numero in entrada.split(",")]

print(lista_enteros)
#print(lista_enteros2)

#Utilizando funciones built-in, determine el número máximo, el número mínimo y la suma total de los números introducidos.
#max()
#min()
#sum()

maximo = max(lista_enteros)
minimo = min(lista_enteros)
suma = sum(lista_enteros)

# 3. Mostrar los resultados
print("\nNúmero máximo:",maximo)
print("\nNúmero mínimo:",minimo)
print("\nSuma total:", suma)

#Determine la verdad o falsedad de la siguiente afirmación: "El número máximo menos el número mínimo es igual a la suma total dividida por 5".
afirmacion = ((maximo - minimo) == (suma / 5))
#la variable afirmacion contendrá TRUE ó FALSE
# combinaciones que SÍ la cumplen:[1, 2, 2][4, 4, 7][5, 6, 9]
print("\nEs",afirmacion,"que el número máximo menos el número mínimo es igual a la suma total dividida por 5")
#Muestre los resultados utilizando print.

print("-" * 32)
```
