## Ejercicio 1
Somos una empresa que tiene varias tiendas de electrodomesticos. Necesitamos un programa para manejar las tiendas, ver las ventas que han tenido, cuántos empleados hay, etc... Para ello vamos a modelizar la tienda en una clase, que tendrá las siguientes características:
* Nombre clase: "Tienda"
* Atributos comunes:
    * Tipo: "Electrodomésticos"
    * Abierta: True
* Atributos propios de cada tienda:
    * Nombre - String
    * Dirección - String
    * Número de empleados - int
    * Ventas ultimos 3 meses - Lista de 3 numeros
* Método para calcular las ventas de todos los meses, que devuelva un numero con todas las ventas.
* Método que calcula la media de ventas de los ultimos meses, por empleado
* Método que devuelve en un string el nombre de la tienda, junto con su dirección.
* Método que obtiene las ventas del último mes.
* Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda. Siendo X el parámetro de entrada. Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento.

Se pide:
1. Implementar la clase "Tienda"
2. Crear tres tiendas con datos inventados
3. Comprueba en al menos una de ellas, todo lo implementado en la clase tienda (sus atributos, media de ventas, ventas/empleado...)
3. Calcular las ventas del último mes de todas las tiendas. Para ello usa el bucle `for`
4. Imprime por pantalla los nombres de las tiendas cuya dirección lleve el string "Avenida"
5. Seguro que ya has hecho este apartado... Documenta la clase :)

```python
class Tienda:
    # Atributos comunes:
    tipo ="Electrodomésticos"
    abierta=True

    def __init__(self, nombre: str, direccion: str, num_empleados: int, ventas_ultimos3meses: list):
        # Atributos propios de cada tienda:
        # Nombre - String
        # Dirección - String
        # Número de empleados - int
        # Ventas últimos 3 meses - Lista de 3 numeros
        self.nombre = nombre
        self.direccion = direccion
        self.num_empleados = num_empleados
        self.ventas_ultimos3meses = ventas_ultimos3meses

    #el símbolo -> se emplea para avisar o indicar anotaciones del tipo de retorno de una función.
    def calcular_total_ventas(self) -> float: 
        # Método para calcular las ventas de todos los meses, que devuelva un numero con todas las ventas.
        # Devuelve la suma total
        return sum(self.ventas_ultimos3meses)

    def calcular_media_por_empleado(self) -> float:
        # Método que calcula la media de ventas de los ultimos meses, por empleado
        if self.num_empleados == 0: 
            # IF para prevenir el ERROR de que NO se puede dividir entre cero
            return 0.0
        # Calcula la media de ventas acumuladas por cada empleado de la tienda.
        total_ventas = self.calcular_total_ventas()
        return total_ventas / self.num_empleados

    def NombreTienda_Direccion(self) -> str:
        # Método que devuelve en un string el nombre de la tienda, junto con su dirección.
        return "La tienda: "+ self.nombre + " está en la dirección: "+ self.direccion

    def ventas_ultimo_mes(self) -> float:
        # Método que obtiene las ventas del último mes = último elemento de la lista.
        return self.ventas_ultimos3meses[-1]

    def proyeccion_ventas_marketing(self, inversion_marketing: float) -> list:
        # Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda. 
        # Siendo X el parámetro de entrada. 
        # Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), 
        # si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). 
        # El método reescribe el atributo de ventas en los últimos 3 meses, 
        # y además devuelve las ventas con el aumento.

        # Determinando el factor según la inversión X
        if inversion_marketing < 1000:
            factor = 1.2 
        else:
            factor = 1.5
        # factor = 1.2 if inversion_marketing < 1000 else 1.5
        
        # Reescribiendo el atributo con el aumento
        self.ventas_ultimos3meses = [venta * factor for venta in self.ventas_ultimos3meses]
        return self.ventas_ultimos3meses


# Creación de tres tiendas con datos inventados
tienda_1 = Tienda("Cacharritos Premium", "Calle Kalea", 3, [100000, 200000, 300000])
tienda_2 = Tienda("Cacharritos Deluxe", "Kalea Avenida", 4, [200000, 300000, 400000])
tienda_3 = Tienda("Outlet Cacharritos", "Avenida Kalea", 5, [300000, 400000, 500000])

todas_las_tiendas = [tienda_1, tienda_2, tienda_3]


# Comprueba en al menos una de ellas, todo lo implementado en la clase tienda 
# (sus atributos, media de ventas, ventas/empleado...)
print("-" * 32)
print("\nComprobaciones de la tienda:", tienda_1.nombre)
# Atributos comunes y propios
print("Tipo de negocio:",tienda_1.tipo)
print("¿Está abierta?:", tienda_1.abierta)
print("Nombre:", tienda_1.nombre)
print("Dirección:", tienda_1.direccion)
print("Empleados:", tienda_1.num_empleados)
print("Ventas de los últimos 3 meses:", tienda_1.ventas_ultimos3meses)

# total de ventas, media de ventas por empleado, ventras del último mes
print("-" * 32)
print("Total ventas (3 meses):", tienda_1.calcular_total_ventas(),"€")
print("Media de ventas por empleado:", tienda_1.calcular_media_por_empleado(),"€")
print("Tienda y Dirección:", tienda_1.NombreTienda_Direccion())
print("Ventas último mes:", tienda_1.ventas_ultimo_mes(),"€")

# Proyección de ventas de marketing: Inversión menor a 1000
print("-" * 32)
print("Ventas de los últimos 3 meses:", tienda_1.ventas_ultimos3meses)
print("Aplicando inversión en marketing de 500€ (Factor 1.2):")
ventas_proyectadas = tienda_1.proyeccion_ventas_marketing(500)
print("Proyección de las ventas según el método:", ventas_proyectadas)
print("Ventas reescritas con el aumento:", tienda_1.ventas_ultimos3meses)
print("-" * 32)

# Calcular las ventas del último mes de todas las tiendas. Para ello usa el bucle for
print("\n\nVentas del último mes en todas las tiendas")
print("-" * 32)

for tienda in todas_las_tiendas:
    print("-" * 32)
    print("Comprobaciones de la tienda:", tienda.nombre)
    print("Tipo de negocio:",tienda.tipo)
    print("¿Está abierta?:", tienda.abierta)
    print("Nombre:", tienda.nombre)
    print("Dirección:", tienda.direccion)
    print("Empleados:", tienda.num_empleados)
    print("Ventas de los últimos 3 meses:", tienda.ventas_ultimos3meses)

    ventas_ultimo_mes = tienda.ventas_ultimo_mes()
    print(tienda.nombre, ": Ventas del último mes", ventas_ultimo_mes, "€")
    print("-" * 32)
print("-" * 32)

# Imprime por pantalla los nombres de las tiendas cuya dirección lleve el string "Avenida"
# Filtrando tiendas por dirección = "Avenida")
print("\nLas tiendas cuya dirección contiene 'Avenida' son:")
for tienda in todas_las_tiendas:
    if "Avenida" in tienda.direccion:
        print(tienda.nombre, "en", tienda.direccion)
print("-" * 32)
```

## Ejercicio 2
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
* Cosas que sabemos seguro que tiene un perro
    * Tiene 4 patas
    * 2 orejas
    * 2 ojos
    * Una velocidad de 0. Por defecto, el perro está parado
* Cuando se inicialice:
    * El perro será de una determinada raza
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario.
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario.
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
* Necesita otro método (parar), donde pondremos la velocidad a 0.
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.


Se pide:
* Implementa la clase *Perro*
* Crea un objeto de tipo *Perro*, sin dueño
* Comprueba que están bien todos sus atributos
* Prueba que ande, y comprueba su velocidad
* Páralo
* Documenta la clase *Perro*

```python
class Perro:
    # Implementando la clase Perro
    # Atributos de clase (comunes para todos los perros)
    patas = 4
    orejas = 2
    ojos = 2
    velocidad = 0 # Una velocidad de 0. Por defecto, el perro está parado
    
    def __init__(self, patas: int, orejas:int , ojos:int, velocidad: int, raza="Podenco", pelo="Marrón", dueño=None ):
        self.patas = patas
        self.orejas = orejas
        self.ojos = ojos
        self.velocidad = velocidad
        self.raza = raza
        self.pelo = pelo
        self.dueño = dueño

    def andar(self, aumento_velocidad):
        # método andar: que tiene un argumento de entrada (aumento_velocidad). 
        # Este valor se le sumará a la velocidad que ya llevaba el perro.
        self.velocidad = self.velocidad + aumento_velocidad

    def parar(self):
        # método parar: donde pondremos la velocidad a 0.
        self.velocidad = 0

    def ladrar(self, mensaje) -> str:
        # método ladrar: tendrá un argumento de entrada, 
        # y la salida será el siguiente string: "GUAU!" + el argumento de entrada.
        return "GUAU! " + mensaje

# Crea un objeto de tipo Perro, sin dueño
# required positional arguments: 'patas', 'orejas', 'ojos', and 'velocidad'
perrete = Perro(4,2,2,3)

print("*" *32)
# Comprueba que están bien todos sus atributos
print("\nComprobación de atributos correctos\n")
print("Raza:", perrete.raza)
print("Color de pelo:", perrete.pelo)
print("Dueño:", perrete.dueño)
print("Patas:", perrete.patas, "\nOrejas:", perrete.orejas, "\nOjos:", perrete.ojos)
print("Velocidad inicial:", perrete.velocidad)

print("*" *32)
# Prueba que ande, y comprueba su velocidad
print("\nPrueba de andar y comprobación de velocidad\n")
# print("Introduce en número:")
# aumento_velocidad= int(input("¿En cuánto aumenta la velocidad al andar?:"))
# print("\nLa velocidad del perrete andando con el aumento de velocidad es de :",perrete.velocidad)

while True:
    entrada = input("¿En cuánto aumenta el perrete la velocidad al andar? Introduce un número entero: ")
    # entrada es tipo string
    try:
        aumento_velocidad = int(entrada)
        # Si la entrada no es numérica, salta al bloque 'except'
        # Si la transformación funciona, el código continúa aquí y rompe el bucle con el BREAK
        perrete.andar(aumento_velocidad)
        print("\nLa velocidad del perrete andando con el aumento de velocidad es de :",perrete.velocidad)
        break
    except ValueError:
        # Previene que el programa se rompa y gestiona el error de datos inválidos
        print("ERROR, entrada no válida. Asegúrate de introducir un número entero válido (dígitos numéricos)")


print("*" *32)

# Páralo
print("\nPrueba de parar y comprobación de velocidad\n")
perrete.parar()
print("Velocidad al parar:", perrete.velocidad)

print("*" *32)
# Prueba del método ladrar
print("\nPrueba de ladrar\n")
# print("El perrete dice: -", perrete.ladrar("Quiero salir a dar el paseo perruno"))
ladrido = input("¿Qué dice el perrete? ")
print("El perrete dice: -", perrete.ladrar(ladrido))

print("*" *32)
```
