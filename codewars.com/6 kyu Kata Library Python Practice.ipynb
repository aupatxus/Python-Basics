````
Write a function that will return the count of distinct case-insensitive alphabetic characters and numeric digits that occur more than once in the input string. The input string can be assumed to contain only alphabets (both uppercase and lowercase) and numeric digits.
Example

"abcde" -> 0 # no characters repeats more than once
"aabbcde" -> 2 # 'a' and 'b'
"aabBcde" -> 2 # 'a' occurs twice and 'b' twice (`b` and `B`)
"indivisibility" -> 1 # 'i' occurs six times
"Indivisibilities" -> 2 # 'i' occurs seven times and 's' occurs twice
"aA11" -> 2 # 'a' and '1'
"ABBA" -> 2 # 'A' and 'B' each occur twice
````

`````python
def duplicate_count(text):
    texto = text.lower()
    visto = []
    duplicados = []
    
    for letra in texto:
        if letra in visto and letra not in duplicados: # Si ya lo habíamos visto antes y aún no está en la lista de duplicados...
            duplicados.append(letra)
        else:
            visto.append(letra)
    print("Analizado:", visto, "en el texto", texto, "Se han encontrado duplicados", duplicados )            
    return len(duplicados)
`````

`````python
print(duplicate_count("abcde")) # 0 # no characters repeats more than once
print(duplicate_count("aabbcde")) # 2 # 'a' and 'b'
print(duplicate_count("aabBcde")) # 2 # 'a' occurs twice and 'b' twice (`b` and `B`)
print(duplicate_count("indivisibility")) # 1 # 'i' occurs six times
print(duplicate_count("Indivisibilities")) # 2 # 'i' occurs seven times and 's' occurs twice
print(duplicate_count("aA11")) # 2 # 'a' and '1'
print(duplicate_count("ABBA")) # 2 # 'A' and 'B' each occur twice
`````

````
Convert string to camel case

Complete the method/function so that it converts dash/underscore delimited words into camel casing. The first word within the output should be capitalized only if the original word was capitalized (known as Upper Camel Case, also often referred to as Pascal case). The next words should be always capitalized.
Examples

"the-stealth-warrior" gets converted to "theStealthWarrior"

"The_Stealth_Warrior" gets converted to "TheStealthWarrior"

"The_Stealth-Warrior" gets converted to "TheStealthWarrior"
````

`````python
def to_camel_case(text):
    
    nuevotexto = text.replace("_", "-") # Reemplazamos todos los guiones bajos por guión medio
    #print(nuevotexto)
    
    palabras = nuevotexto.split("-") # Empleando el guión medio como separador, separamos el texto en palabras dentro de una lista
    #print (palabras)

    resultado = palabras[0] # Guardamos la primera palabra tal cual venga

    for palabra in palabras[1:]: # Procesamos el resto de palabras empleando upper() en su primera letra
        resultado = resultado + palabra[0].upper() + palabra[1:].lower()
            
    return resultado



`````

````
Per Wikipedia, a prime number ( or a prime ) is a natural number greater than 1 that has no positive divisors other than 1 and itself.
Requirements

    You can assume you will be given an integer input.
    You can not assume that the integer will be only positive. You may be given negative numbers as well ( or 0 ).
    NOTE on performance: There are no fancy optimizations required, but still the most trivial solutions might time out. Numbers go up to 2^31 ( or similar, depending on language ). Looping all the way up to n, or n/2, will be too slow.

Example

is_prime(1)  /* false */
is_prime(2)  /* true  */
is_prime(-1) /* false */

````

`````python
def is_prime(num):
    
    if num <= 1: # No son primos los números (negativos, el cero) menores o iguales a 1
        return False
    
    if num <= 3: # El 2 y el 3 son primos
        return True
    
    if num % 2 == 0 or num % 3 == 0: # No son primos múltiplos de 2 y de 3
        return False
    
    for i in range(5, num, 2): # buscamos con números impares (3, 5, 7, 9...)
        if i * i > num: # Si el cuadrado de 'i' ya es mayor que 'n', paramos de buscar
            break
        
        if num % i == 0: # no es primo si num se puede dividir exactamente por i
            return False
            
    return True


`````

````
Instead, if I try to go further up from the upmost or further down from the downmost, I'll just stay where I am located (examples: you can't go lower than lowest row: Ken, Chun Li, Zangief, Dhalsim, Sagat and M.Bison in the above image; you can't go upper than highest row: Ryu, E.Honda, Blanka, Guile, Balrog and Vega in the above image).
````

`````python
def street_fighter_selection(fighters, initial_position, moves):
   
    # Guardamos la posición actual separando fila (y) y columna (x)
    fila, columna = initial_position
    personajes_recorridos = []
    
    # Dimensiones de la cuadrícula (en este caso 2 filas y 6 columnas)
    filas = len(fighters)
    columnas = len(fighters[0])
    
    for movimiento in moves:
        # Movimiento Vertical (Bloqueado entre la fila 0 y la última fila)
        if movimiento == 'up':
            fila = max(0, fila - 1) #max(0, fila - 1) asegura que si el resultado es negativo, se quede clavado en 0.
        elif movimiento == 'down':
            fila = min(filas - 1, fila + 1) # min(filas - 1, fila + 1) asegura que nunca pase de la última fila disponible.
        
        # Movimiento Horizontal (Circular, da la vuelta usando %)
        # Si estás en la columna 0 y vas a la izquierda (0 - 1 = -1), al hacer -1 % 6 el resultado es 5 (la última columna). 
        # Si estás en la 5 y vas a la derecha (5 + 1 = 6), 6 % 6 vuelve a ser 0 (la primera columna).
        elif movimiento == 'left':
            columna = (columna - 1) % columnas
        elif movimiento == 'right':
            columna = (columna + 1) % columnas
            
        # Añadimos el personaje que está bajo el cursor tras el movimiento
        personaje_actual = fighters[fila][columna]
        personajes_recorridos.append(personaje_actual)
        
    return personajes_recorridos
`````

````
Dashatize it
Given an integer, return a string with dash '-' marks before and after each odd digit, but do not begin or end the string with a dash mark.

Ex:

274 -> '2-7-4'
6815 -> '68-1-5'

````

`````python

def dashatize(n):

    num_string = str(n)
    resultado = []
    resultado_string = ""
    
    for i in range(len(num_string)):
        if num_string[i] != "-":
            resultado.append(num_string[i])
        
        if i < len(num_string) - 1 and num_string[i] != "-": # Si NO es el último dígito, miramos el dígito actual y el que viene después
            actual = int(num_string[i])
            siguiente = int(num_string[i+1])
            
            # Si el actual ó el siguiente son impares, hay que poner un guion medio
            if actual % 2 != 0 or siguiente % 2 != 0:
                resultado.append("-")
    
    resultado_string = "".join(resultado)
    return resultado_string 
`````

`````python
def what_century(year):
    resultado = ""
    año = int(year) # texto a número entero
    siglo = (año - 1) // 100 + 1 # Restamos 1 para que años como el 2000 caigan en el siglo 20 (1999 // 100 + 1 = 20)
    print("El año", año, "corresponde al siglo", siglo)
    
    # terminación en inglés: th, st, nd, rd
    if 11 <= siglo <= 13: # los siglos 11, 12 y 13 siempre terminan en "th"
        terminacion = "th"
    else:
        ultimo_digito = siglo % 10
        if ultimo_digito == 1:
            terminacion = "st"
        else:
            if ultimo_digito == 2:
                terminacion = "nd"
            else:
                if ultimo_digito == 3:
                    terminacion = "rd"
                else:
                    terminacion = "th"
            
    resultado = str(siglo) + terminacion
    return resultado
`````
