````
String will never be empty and you do not need to account for different data types.
````

`````python
def find_short(s):
    return min(len(word) for word in s.split())
#  s.split(): Divide la cadena de texto en una lista de palabras individuales, eliminando automáticamente los espacios en blanco.
# (len(word) for word in s.split()): Calcula la longitud (len) de cada palabra dentro de esa lista.
# min(...): Analiza todos los tamaños y devuelve el número más pequeño, que corresponde a la longitud de la palabra más corta.
`````

`````python
find_short("bitcoin take over the world maybe who knows perhaps")
`````

`````python
def find_short(s):
    # Separamos la cadena de palabras en una lista 
    list_palabras = s.split()
    # print(list_palabras)
    
    # Inicialización de la variable con la longitud de la primera palabra
    longitud_mas_corta = len(list_palabras[0])
    
    for palabra in list_palabras:
        # Si la palabra actual es más corta que la anterior...
        if len(palabra) < longitud_mas_corta:
            # Actualizamos con la nueva longitud más corta
            longitud_mas_corta = len(palabra)
            
    return longitud_mas_corta
`````

````
Jaden-Cased:     "How Can Mirrors Be Real If Our Eyes Aren't Real"

````

`````python
def to_jaden_case(string):
    return " ".join(word.capitalize() for word in string.split(" "))
# string.split(" "): Divide la frase en palabras individuales utilizando los espacios en blanco.
# word.capitalize(): Este método convierte estrictamente solo la primera letra de cada palabra en mayúscula y mantiene las demás en minúscula
# " ".join(...): Vuelve a unir todas las palabras procesadas metiendo un espacio entre ellas para reconstruir la frase original estilo Jaden Case.
`````

`````python
def to_jaden_case(string):
    resultado = []
    for word in string.split(" "):
        # print(word)
        resultado.append(word.capitalize())
    # print(resultado)
    return " ".join(resultado)
`````

````
// "What was the name of your first pet?"
````

`````python
def maskify(cc):
    # Si la cadena tiene 4 caracteres o menos, se devuelve tal cual
    if len(cc) <= 4:
        return cc
    else:
        # Se multiplica # por la cantidad de caracteres a ocultar
        # y se suma los últimos 4 caracteres usando un índice negativo [-4:]
        return ("#" * (len(cc) - 4)) + cc[-4:]


`````

`````python
def validate_pin(pin):
    """
    Returns True if pin is a string of 4 or 6 digits, False otherwise.
    """
    
    # First check that pin is a string of length 4 or 6
    if(type(pin) != str or len(pin) not in [4, 6]):
        return(False)

    # If any character is not a digit, return False
    for c in pin:
        if c not in "0123456789":
            return(False)

    # If all the characters are digits, return True
    return(True)
`````

`````python
def reverse_words(text):
    palabras_invertidas = []
    list_palabras = text.split(" ") 
    
    for palabra in list_palabras:
        # Invertimos la palabra completa y la guardamos en la lista
        palabras_invertidas.append(palabra[::-1])
        
    # Unimos todas las palabras invertidas usando un espacio como separador
    return " ".join(palabras_invertidas)
`````

````
Note: anagrams are case insensitive

Complete the function to return true if the two arguments given are anagrams of each other; return false otherwise.
Examples

    "foefet" is an anagram of "toffee"

    "Buckethead" is an anagram of "DeathCubeK"

````

`````python
def is_anagram(test, original):
    # no pueden ser anagramas si no tienen la misma longitud
    if len(test) != len(original):
        return False
    
    # convertimos ambas entradas a minúsculas
    test_minus = test.lower()
    original_minus = original.lower()
    
    # Comprobamos que cada letra esté presente en ambas palabras el mismo número de veces
    for letra in test_minus:
        if test_minus.count(letra) != original_minus.count(letra):
            return False
            
    return True
`````
